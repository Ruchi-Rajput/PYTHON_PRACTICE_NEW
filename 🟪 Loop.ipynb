```python
# 3-step trangle
n = 5
for i in range(1,n):         # i = 1,2,3,4 (stops before n)
    for j in range(n-i):     # print left padding:"-",repeated(n-i)times 
        print("-",end=" ")   
    for k in range(i):       # print the left side + center:"x" repeated i times
        print("x",end =" ")   
    for l in range(1,i):     # print the right side: "x" repeated(i-1) times
        print("x",end =" ")   
    print()                    # move to the next
```

```python
# printing the maximum and minimum element in given list without using built-in function:--
List_1 = [12,3,4,56,78,90,90,12,23]
max_e = List_1[0]
for element in List_1:
    if element < max_e:
        max_e = element
print(max_e) 
```

```python
# Average of list without using built-in function:--
List_1 = [12,3,4,56,78,90,90,12,23]

sum_num = 0
count_num = 0

for element in List_1:
    sum_num += element
    count_num += 1
print(f"Sum of list elements: {sum_num}")
print(f"Count of list elements: {count_num}")

Average = sum_num/count_num
print(f"Average of list elements: {Average}")
```
