```python
# copy to previous class 
def prime(n):
    if n>1:
        for i in range(2,n):
            if n%i == 0:
                print("Not prime")
                return n
                break
        else:
            print("Prime")
            return n
    else:
        print("Invalid Number")
prime(55)
```

```python
def prime(n):
    if n == 0 or n == 1:
        print("Not a prime number")
    elif n > 2:
        for i in range(2,n):
            if n%i == 0:
                print(n,"-Not prime number")
                break
        else:
            print(n,"-Prime number")
    else:
        print("Invalid number")
prime(33)
```

```python
### maximum of two numbers:-
def maxx_2(a,b):  # with argument with return type
    if a>b:
        return a
    else:
        return b
maxx_2(4,8)
```

```python
# maximum of three numbers:-
def maxx_3(a,b,c):  # with argument with return type
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c
maxx_3(8,9,5)
```

```python
# Factorial numbers:-
def factorial(n):
    if n == 1:
        return n
    else:
        return n*factorial(n-1)  
factorial(10)
```
