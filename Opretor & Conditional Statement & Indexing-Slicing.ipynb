<h4 style="color:brown">Slicing</h4>

```python
lst_1 = [45,67,89,90,45,23,14,56]
print(lst_1[0:3])  # Here we provide Starting(included) and ending(not included) position 
```

```python
# sub-set of list
# In slicing we give starting position and ending or stepsize
```

```python
print(lst_1[ :5])  # Here we provid only ending position it get starting position by default.
```

```python
print(lst_1[:]) # Here it can take start,end both position by default.
```

```python
lst_1 = [45,67,89,[90,45,23],14,56]
print(lst_1[:4]) 
print(lst_1[3][2])
```

```python
lst_1s = [45,67,89,[90,45,23],14,56]
```

```python
print(lst_1s[0:4:2]) # It's skip the value number that we put as stepsize 
print(lst_1s[0:6:2])
print(lst_1s[::2]) 
print(lst_1s[::]) # Here it's take all three position by default stepsize by default +1
```

```python
print(lst_1s[-1:-4])  # No output becausse stepsize also show direction
# and here direction is by default +1 ,
# but we are fetching negative value and going in positive direction
```

```python
print(lst_1s[-1:-4:-1]) # here updated it and provide right direction.
```

```python
print(lst_1s[-1:-7:-1])
```
